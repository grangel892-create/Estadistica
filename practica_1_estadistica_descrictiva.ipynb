## Realizar Test de Shapiro-Wilk para Peso y Talla

### Subtask:
Aplicar el test de Shapiro-Wilk a las variables 'PESO (Gramos)' y 'TALLA (Centímetros)' para evaluar su normalidad. Este test es adecuado para muestras pequeñas a moderadas y devuelve un estadístico y un valor p.


**Reasoning**:
To perform the Shapiro-Wilk test, I need to import the `shapiro` function from the `scipy.stats` module and then apply it to the specified columns, printing the statistic and p-value for each.



```python
from scipy.stats import shapiro

# Apply Shapiro-Wilk test to 'PESO (Gramos)'
shapiro_peso_stat, shapiro_peso_p = shapiro(df['PESO (Gramos)'])
print(f"Shapiro-Wilk Test for PESO (Gramos):\n Statistic={shapiro_peso_stat:.3f}, p-value={shapiro_peso_p:.3f}")

# Apply Shapiro-Wilk test to 'TALLA (Centímetros)'
shapiro_talla_stat, shapiro_talla_p = shapiro(df['TALLA (Centímetros)'])
print(f"Shapiro-Wilk Test for TALLA (Centímetros):\n Statistic={shapiro_talla_stat:.3f}, p-value={shapiro_talla_p:.3f}")
```

## Realizar Test de Kolmogorov-Smirnov para Peso y Talla

### Subtask:
Aplicar el test de Kolmogorov-Smirnov a las variables 'PESO (Gramos)' y 'TALLA (Centímetros)' para evaluar su normalidad. Este test compara la distribución observada con una distribución teórica (normal) y devuelve un estadístico D y un valor p.


**Reasoning**:
To perform the Kolmogorov-Smirnov test for normality, I will first standardize the 'PESO (Gramos)' column by subtracting its mean and dividing by its standard deviation. Then, I will apply the `kstest` function from `scipy.stats` to compare the standardized data against a standard normal distribution.



```python
from scipy.stats import kstest
from scipy.stats import norm

# Standardize 'PESO (Gramos)' for the Kolmogorov-Smirnov test
mean_peso = df['PESO (Gramos)'].mean()
std_peso = df['PESO (Gramos)'].std()
standardized_peso = (df['PESO (Gramos)'] - mean_peso) / std_peso

# Apply Kolmogorov-Smirnov test to 'PESO (Gramos)'
ks_peso_stat, ks_peso_p = kstest(standardized_peso, 'norm')
print(f"Kolmogorov-Smirnov Test for PESO (Gramos):\n Statistic={ks_peso_stat:.3f}, p-value={ks_peso_p:.3f}")
```

**Reasoning**:
To complete the subtask, I will now apply the Kolmogorov-Smirnov test to the 'TALLA (Centímetros)' column. This involves standardizing the column first, then using `kstest` to compare it against a normal distribution, and finally printing the results.



```python
from scipy.stats import kstest
from scipy.stats import norm

# Standardize 'TALLA (Centímetros)' for the Kolmogorov-Smirnov test
mean_talla = df['TALLA (Centímetros)'].mean()
std_talla = df['TALLA (Centímetros)'].std()
standardized_talla = (df['TALLA (Centímetros)'] - mean_talla) / std_talla

# Apply Kolmogorov-Smirnov test to 'TALLA (Centímetros)'
ks_talla_stat, ks_talla_p = kstest(standardized_talla, 'norm')
print(f"Kolmogorov-Smirnov Test for TALLA (Centímetros):\n Statistic={ks_talla_stat:.3f}, p-value={ks_talla_p:.3f}")
```
